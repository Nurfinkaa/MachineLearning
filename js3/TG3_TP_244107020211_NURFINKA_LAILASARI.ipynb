# **Tugas Praktikum**

Wisconsin Breast Cancer


### **Deskripsi Tugas**

Pada tugas pratikum ini Anda akan menggunakan data "Wisconsin Breast Cancer". Data tersebut terdiri dari 569 data yang digunakan untuk mendiagnonis jenis kanker Malignant (M) dan Benign (B). Tugas Anda adalah,

1. Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.

2. Lakukan proses encoding pada kolom "diagnosis".

3. Lakukan proses standarisasi pada semua kolom yang memiliki nilai numerik.

4. Lakukan proses stratified split data untuk membuat data latih dan data uji dengan rasio 80:20.

### **Langkah 0 - Upload & Load Data**

```python
import pandas as pd
from google.colab import files
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
```

```python
uploaded = files.upload()
dpath = list(uploaded.keys())[0]
```

```python
df = pd.read_csv(dpath)
print("Ukuran data:", df.shape)
```

```python
display(df.head())
```

```python
df.info()
```

### **Langkah 1 - Pisahkan variabel yang dapat dan tidak dapat digunakan**

```python
# 'id' hanya identitas pasien, 'Unnamed: 32' kolom kosong (jika ada)
kolom_tidak_dipakai = [c for c in ['id', 'Unnamed: 32'] if c in df.columns]
print("Kolom yang tidak dapat digunakan:", kolom_tidak_dipakai)
```

```python
df = df.drop(columns=kolom_tidak_dipakai, errors='ignore')
print("Kolom yang digunakan:", list(df.columns))
display(df.head())
```

### **Langkah 2 - Encoding kolom “diagnosis”**

```python
le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])
```

```python
print("Mapping encoding:", dict(zip(le.classes_, le.transform(le.classes_))))
print(df['diagnosis'].value_counts())
display(df.head())
```

### **Langkah 3 - Standarisasi kolom numerik**

```python
# kolom 'diagnosis' (target) tidak ikut distandarisasi
fitur_numerik = df.select_dtypes(include='number').columns.drop('diagnosis')

scaler = StandardScaler()
df[fitur_numerik] = scaler.fit_transform(df[fitur_numerik])
```

```python
print("=== Hasil standarisasi (mean ≈ 0, std ≈ 1) ===")
display(df[fitur_numerik].describe().loc[['mean', 'std']].round(3))
```

```python
display(df.head())
```

### **Langkah 4 - Stratified split 80:20**

```python
X = df.drop(columns='diagnosis')
y = df['diagnosis']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```

```python
print("Jumlah data latih :", X_train.shape[0])
```

```python
print("Jumlah data uji   :", X_test.shape[0])
```

```python
print("\nProporsi kelas data latih:\n", y_train.value_counts(normalize=True).round(3))
```

```python
print("\nProporsi kelas data uji:\n", y_test.value_counts(normalize=True).round(3))
```
