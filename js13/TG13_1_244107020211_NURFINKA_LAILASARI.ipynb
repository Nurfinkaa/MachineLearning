```python
import numpy as np

# Dataset XOR
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

# Parameter
input_size = 2
hidden_size = 2
output_size = 1
lr = 0.1

# Inisialisasi bobot
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# Fungsi aktivasi
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Training
for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # Hitung error
    error = y - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss}")

# Output akhir
print("Prediksi:")
print(a2)
```

```python
import numpy as np

# =========================
# Dataset XOR
# =========================
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [1], [1], [0]])

# =========================
# Fungsi aktivasi
# =========================
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(a):
    # a = sigmoid(z)
    return a * (1 - a)

def relu(x):
    return np.maximum(0, x)

def relu_derivative(z):
    # z = pre-aktivasi
    return (z > 0).astype(float)

# =========================
# Fungsi training
# =========================
def train(hidden_size, activation="sigmoid", epochs=10000, lr=0.1,
          seed=42, verbose=True):
    rng = np.random.RandomState(seed)
    input_size, output_size = 2, 1

    # Inisialisasi bobot dan bias
    W1 = rng.randn(input_size, hidden_size)
    b1 = np.zeros((1, hidden_size))
    W2 = rng.randn(hidden_size, output_size)
    b2 = np.zeros((1, output_size))

    for epoch in range(epochs):
        # Forward pass
        z1 = np.dot(X, W1) + b1
        a1 = sigmoid(z1) if activation == "sigmoid" else relu(z1)
        z2 = np.dot(a1, W2) + b2
        a2 = sigmoid(z2)  # output layer tetap sigmoid

        # Hitung error
        error = y - a2

        # Backpropagation
        d_a2 = error * sigmoid_derivative(a2)
        d_W2 = np.dot(a1.T, d_a2)
        d_b2 = np.sum(d_a2, axis=0, keepdims=True)

        if activation == "sigmoid":
            d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
        else:
            d_a1 = np.dot(d_a2, W2.T) * relu_derivative(z1)
        d_W1 = np.dot(X.T, d_a1)
        d_b1 = np.sum(d_a1, axis=0, keepdims=True)

        # Update bobot (gradient descent)
        W1 += lr * d_W1
        b1 += lr * d_b1
        W2 += lr * d_W2
        b2 += lr * d_b2

        if verbose and epoch % 1000 == 0:
            loss = np.mean(np.square(error))
            print(f"Epoch {epoch}, Loss: {loss:.6f}")

    # Forward pass akhir dengan bobot terakhir
    a1 = sigmoid(np.dot(X, W1) + b1) if activation == "sigmoid" \
        else relu(np.dot(X, W1) + b1)
    a2 = sigmoid(np.dot(a1, W2) + b2)
    final_loss = np.mean(np.square(y - a2))
    return final_loss, a2


# =========================
# Eksperimen (seed sama = 42)
# =========================
configs = [
    ("Hidden 2 + Sigmoid (awal)", 2, "sigmoid"),
    ("Hidden 3 + Sigmoid",        3, "sigmoid"),
    ("Hidden 2 + ReLU",           2, "relu"),
    ("Hidden 3 + ReLU",           3, "relu"),
]

results = {}
for name, h, act in configs:
    print("=" * 50)
    print(name)
    print("=" * 50)
    loss, pred = train(h, act, seed=42)
    results[name] = loss
    print("Prediksi:")
    print(np.round(pred, 4))
    print(f"Prediksi (dibulatkan): {pred.round().ravel().astype(int)}")
    print()

# =========================
# Ringkasan perbandingan (seed 42)
# =========================
print("=" * 50)
print("RINGKASAN LOSS AKHIR (seed 42)")
print("=" * 50)
for name, loss in results.items():
    print(f"{name:<28} : {loss:.6f}")

# =========================
# Perbandingan lebih adil: rata-rata 10 seed
# =========================
print()
print("=" * 50)
print("RATA-RATA LOSS AKHIR (10 seed berbeda)")
print("=" * 50)
for name, h, act in configs:
    losses = [train(h, act, seed=s, verbose=False)[0] for s in range(10)]
    print(f"{name:<28} : mean={np.mean(losses):.6f}, "
          f"min={np.min(losses):.6f}, max={np.max(losses):.6f}")
```
