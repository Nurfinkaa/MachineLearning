# **Praktikum 2**

```python
import numpy as np
import tensorflow as tf
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import classification_report, confusion_matrix

# Seed agar hasil bisa direproduksi
np.random.seed(42)
tf.random.set_seed(42)

# =========================
# Load dataset
# =========================
iris = load_iris()
X = iris.data
y = iris.target.reshape(-1, 1)

# =========================
# One-hot encoding
# (sklearn >= 1.2 memakai sparse_output, versi lama memakai sparse)
# =========================
try:
    encoder = OneHotEncoder(sparse_output=False)
except TypeError:
    encoder = OneHotEncoder(sparse=False)
y_onehot = encoder.fit_transform(y)

# =========================
# Split data
# =========================
X_train, X_test, y_train, y_test = train_test_split(
    X, y_onehot, test_size=0.2, random_state=42, stratify=y
)

# =========================
# Bangun model
# =========================
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(4,)),
    tf.keras.layers.Dense(10, activation='relu'),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])

model.summary()

# =========================
# Kompilasi
# =========================
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# =========================
# Latih model
# =========================
history = model.fit(
    X_train, y_train,
    epochs=50,
    batch_size=8,
    validation_split=0.1,
    verbose=1
)

# =========================
# Evaluasi
# =========================
loss, acc = model.evaluate(X_test, y_test, verbose=0)
print(f"\nLoss  : {loss:.4f}")
print(f"Akurasi: {acc:.4f}")

# =========================
# Evaluasi detail
# =========================
y_pred = np.argmax(model.predict(X_test, verbose=0), axis=1)
y_true = np.argmax(y_test, axis=1)

print("\nConfusion Matrix:")
print(confusion_matrix(y_true, y_pred))

print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=iris.target_names))
```

```python
import numpy as np
import tensorflow as tf
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder

# =========================
# Persiapan data (dipakai Tugas 2 dan 3)
# =========================
iris = load_iris()
X = iris.data
y = iris.target.reshape(-1, 1)

# sklearn >= 1.2 memakai sparse_output, versi lama memakai sparse
try:
    encoder = OneHotEncoder(sparse_output=False)
except TypeError:
    encoder = OneHotEncoder(sparse=False)
y_onehot = encoder.fit_transform(y)

X_train, X_test, y_train, y_test = train_test_split(
    X, y_onehot, test_size=0.2, random_state=42, stratify=y
)


def build_and_train(hidden_layers, activation="relu", epochs=50,
                    batch_size=8, seed=42):
    """Bangun, latih, dan evaluasi model. Mengembalikan (loss, akurasi)."""
    tf.keras.backend.clear_session()
    np.random.seed(seed)
    tf.random.set_seed(seed)

    layers = [tf.keras.layers.Input(shape=(4,))]
    for n in hidden_layers:
        layers.append(tf.keras.layers.Dense(n, activation=activation))
    layers.append(tf.keras.layers.Dense(3, activation="softmax"))

    model = tf.keras.Sequential(layers)
    model.compile(optimizer="adam",
                  loss="categorical_crossentropy",
                  metrics=["accuracy"])
    model.fit(X_train, y_train, epochs=epochs, batch_size=batch_size,
              verbose=0)
    loss, acc = model.evaluate(X_test, y_test, verbose=0)
    return loss, acc
```

```python
# =====================================================
# TUGAS 2: Ubah jumlah neuron hidden layer
#          dan bandingkan akurasi dengan konfigurasi awal
# =====================================================
print("=" * 60)
print("TUGAS 2: Variasi jumlah neuron hidden layer (aktivasi ReLU)")
print("=" * 60)

configs_tugas2 = [
    ("Awal (10, 8)",   [10, 8]),
    ("Kecil (4, 3)",   [4, 3]),
    ("Sedang (16, 8)", [16, 8]),
    ("Besar (32, 16)", [32, 16]),
    ("Besar (64, 32)", [64, 32]),
]

hasil_tugas2 = []
for nama, hidden in configs_tugas2:
    loss, acc = build_and_train(hidden, activation="relu")
    hasil_tugas2.append((nama, loss, acc))

print(f"{'Konfigurasi':<18}{'Loss':>10}{'Akurasi':>12}{'Selisih vs awal':>20}")
print("-" * 60)
acc_awal = hasil_tugas2[0][2]
for nama, loss, acc in hasil_tugas2:
    selisih = acc - acc_awal
    print(f"{nama:<18}{loss:>10.4f}{acc:>12.4f}{selisih:>+20.4f}")
```

```python
# =====================================================
# TUGAS 3: Bandingkan Sigmoid vs ReLU pada dataset Iris
#          dan catat perbedaan loss dan akurasi
# =====================================================
print()
print("=" * 60)
print("TUGAS 3: Sigmoid vs ReLU (hidden layer 10, 8)")
print("=" * 60)

hasil_tugas3 = {}
for act in ["sigmoid", "relu"]:
    loss, acc = build_and_train([10, 8], activation=act)
    hasil_tugas3[act] = (loss, acc)

print(f"{'Aktivasi':<12}{'Loss':>10}{'Akurasi':>12}")
print("-" * 34)
for act, (loss, acc) in hasil_tugas3.items():
    print(f"{act:<12}{loss:>10.4f}{acc:>12.4f}")

d_loss = hasil_tugas3["relu"][0] - hasil_tugas3["sigmoid"][0]
d_acc = hasil_tugas3["relu"][1] - hasil_tugas3["sigmoid"][1]
print(f"\nSelisih (ReLU - Sigmoid): loss={d_loss:+.4f}, akurasi={d_acc:+.4f}")
```

```python
# Perbandingan lebih adil: rata-rata 5 seed
print()
print("=" * 60)
print("TUGAS 3 (tambahan): Rata-rata 5 seed berbeda")
print("=" * 60)
print(f"{'Aktivasi':<12}{'Mean Loss':>12}{'Mean Akurasi':>16}{'Std Akurasi':>14}")
print("-" * 54)
for act in ["sigmoid", "relu"]:
    hasil = [build_and_train([10, 8], activation=act, seed=s) for s in range(5)]
    losses = [h[0] for h in hasil]
    accs = [h[1] for h in hasil]
    print(f"{act:<12}{np.mean(losses):>12.4f}{np.mean(accs):>16.4f}{np.std(accs):>14.4f}")
```
